```python
from pypdf import PdfReader
from pathlib import Path

PDF_FOLDER = Path.cwd() / "pdfs"
pdf_path = list(PDF_FOLDER.glob("*.pdf"))[0]

reader = PdfReader(str(pdf_path))

print("PDF:", pdf_path.name)
print("Total pages:", len(reader.pages))

text = "\n".join(
    page.extract_text() or ""
    for page in reader.pages
)

print("\nFirst 1000 characters:\n")
print(text[:1000])
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=800,
    chunk_overlap=150
)

chunks = text_splitter.split_text(text)

print("Total chunks:", len(chunks))
print("\nFirst chunk:\n")
print(chunks[0])
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

print("Embedding model loaded successfully!")
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma.from_texts(
    texts=chunks,
    embedding=embeddings,
    collection_name="aifod_pdf"
)

print("Vector database created successfully!")
print("Total chunks stored:", vectorstore._collection.count())
```

```python
query = "What are the main features of the AIFOD AI Interview Platform?"

results = vectorstore.similarity_search(query, k=3)

for i, result in enumerate(results, start=1):
    print(f"\n--- Result {i} ---")
    print(result.page_content)
```

```python
def ask_pdf(question):
    # Find relevant information in your PDF
    results = vectorstore.similarity_search(question, k=3)

    # Combine the retrieved text
    context = "\n\n".join(
        doc.page_content for doc in results
    )

    # Ask Gemini to answer using the PDF
    prompt = f"""
    Answer using only the context from the PDF.
    If the answer is not available, say so.

    Context:
    {context}

    Question: {question}
    """

    response = llm.invoke(prompt)
    return response.content


print("PDF chatbot function created successfully!")
```

```python
def get_clean_text(response):
    content = response.content

    if isinstance(content, str):
        return content

    if isinstance(content, list):
        return "\n".join(
            item["text"]
            for item in content
            if isinstance(item, dict) and "text" in item
        )

    return str(content)

print("Output-cleaning function created successfully!")
```
